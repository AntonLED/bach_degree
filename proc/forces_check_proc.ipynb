```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.genfromtxt(f"../data/forces_check/data.txt", delimiter="\t")

rbot = data[:, 0]

fbt_x = data[:, 1]
fbt_y = data[:, 2]
fbt_z = data[:, 3]

ftb_x = data[:, 4]
ftb_y = data[:, 5]
ftb_z = data[:, 6]

###

plt.figure()
plt.xlabel("rbot")
plt.ylabel("ftb")
plt.grid()
plt.plot(rbot, ftb_x, "-", label="ftb_x")
plt.plot(rbot, ftb_y, "-", label="ftb_y")
plt.plot(rbot, ftb_z, "-", label="ftb_z")
plt.legend(loc="best")
plt.show()

plt.figure()
plt.xlabel("rbot")
plt.ylabel("fbt")
plt.grid()
plt.plot(rbot, fbt_x, "-", label="fbt_x")
plt.plot(rbot, fbt_y, "-", label="fbt_y")
plt.plot(rbot, fbt_z, "-", label="fbt_z")
plt.legend(loc="best")
plt.show()

plt.figure()
plt.xlabel("rbot")
plt.ylabel("delta")
plt.grid()
plt.plot(rbot, fbt_x + ftb_x, "-", label="delta_x")
plt.plot(rbot, fbt_y + ftb_y, "-", label="delta_y")
plt.plot(rbot, fbt_z + ftb_z, "-", label="delta_z")
plt.legend(loc="best")
plt.show()
```
